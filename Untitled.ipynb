```python
import pandas as pd
```

```python
df = pd.read_csv("data/cache/fund_history_value/013840/danjuan_nav_1y.csv")
```

```python
df.date
```

```python
df.head()
```

```python
df["最高点"] = df["value"].cummax()
df["回撤"] = df["value"] / df["最高点"] - 1
df["回撤"] = df["回撤"] * 100
df.head()
```

```python
# 当前是否回撤超过10%
df["below_10"] = df["回撤"] <= -10

# 只标记第一次跌破10%的那一天
df["buy_signal"] = df["below_10"] & (~df["below_10"].shift(1).fillna(False))

buy_points = df[df["buy_signal"]]
buy_points
```

```python
df["MA5"] = df["value"].rolling(window=5).mean()
df["MA10"] = df["value"].rolling(window=10).mean()
df["MA20"] = df["value"].rolling(window=20).mean()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(24, 12))
plt.plot(df["date"], df["value"], label="收盘价")
plt.plot(df["date"], df["MA5"], label="5日线")
plt.plot(df["date"], df["MA10"], label="10日线")

plt.scatter(
    buy_points["date"],
    buy_points["value"],
    s=120,
    marker="o",
    label="首次回撤超过10%",
    zorder=5
)

plt.xlabel("日期")
plt.ylabel("价格")
plt.title("收盘价、5日线和10日线")
plt.legend()
plt.grid(False)
plt.xticks(rotation=45)
plt.savefig("ma5_ma10.svg")
plt.show()
```
